```python
from rustypus import Problem, NSGAII, Real, Integer, BitBinary
```

```python
def my_objective(x):
    f1 = x[0]**2 + x[1]**2
    f2 = (x[0])**2 + (x[1])**2
    return [f1, f2]
```

```python
problem = Problem(
    solution_length=2,
    number_of_objectives=2,
    solution_data_types=[Real(-5.0, 5.0), Real(-5.0, 5.0)],
    objective_function=my_objective,
    direction=[-1, -1],  # -1 = minimize, 1 = maximize
)
```

```python
algo = NSGAII(problem, population_size=10, execution_mode="sequential")
```

```python
%timeit algo.run(10)
```

```python
algo2 = NSGAII(problem, population_size=10, execution_mode="multithreaded")
```

```python
%timeit algo2.run(10)
```

```python
algo3 = NSGAII(problem, population_size=10, execution_mode="gpu")
```

```python
%timeit algo3.run(10)
```
